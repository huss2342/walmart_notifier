# Walmart item history audit

## tl;dr
The October 21 capture contains 70,399 IDs; 95.7% are below $40 and only 18 are $200+. These are observed IDs including variants, not verified claim opportunities.

## Context & Methods
Replay item events over the snapshot, exclude marker events, and count one row per Walmart ID. Use half-open USD bands. The latest history record defines trailing windows.

### Key Assumptions
This local retained history is useful for a pilot's routing design, not a complete Walmart catalog. Re-execution uses current files and can change results. No Walmart, Discord, or Telegram calls occur.

## Data
Inputs: repository data/seen.json and data/seen.log. Exact hashes are in item-history-summary.json. Python standard library only; run from this repository or tasks/research.

## Results

```python
import json
import runpy
from pathlib import Path

root = Path.cwd().resolve()
while not (root / "src/state.py").is_file():
    if root == root.parent:
        raise RuntimeError("Run inside the walmart_notifier repository")
    root = root.parent
audit = runpy.run_path(str(root / "tasks/research/analyze_item_history.py"))
result = audit["analyze"](root)
print(
    json.dumps(
        {
            k: result[k]
            for k in [
                "item_ids",
                "known_values",
                "zero_value_ids",
                "explicit_variant_ids",
                "bands",
                "max_value_usd",
                "trailing_days_ending_at_latest_record",
            ]
        },
        indent=2,
    )
)
```
